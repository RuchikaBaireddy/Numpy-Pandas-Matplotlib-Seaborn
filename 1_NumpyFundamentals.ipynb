```python
# ================= Array Functions ===========================
```

```python
a1=np.random.random((3,3))
a1= np.round(a1*100)
a1
```

```python
# 1) max/min/sum/product
# the follwing gives max/min/sum/product of the entire array
print(np.max(a1))
print(np.min(a1))
print(np.sum(a1)) # adds all the elements in the array and gives its sum
print(np.prod(a1)) # multiplies all the elements in the array and gives its product
```

```python
# if we want  max/min/sum/product of a row/column in an array, we use axis parameter. 
# axis=0 -> column & axis=1 -> row
print(np.max(a1,axis=1)) # gives max value in each row
print(np.max(a1,axis=0)) # gives max value in each column 
```

```python
# # 2) mean/median/std/var. Note: axis parameter can be used for these functions as well.
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
```

```python
# 3)  trignometric functions : all trignometric functions can be used .
np.sin(a1)
```

```python
 # 4)dot product functions ( very useful) - dot product between two arrays is only possible when column of the first array and row of second array match.
a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)
np.dot(a2,a3)
```

```python
# 5) log and exponents
print(np.log(a1))
print(np.exp(a1))
```

```python
# 6) round/floor/ceil
arr = np.random.random((3,3))*100
print(arr)
print(np.round(arr)) # rounds off to the nearest number
print(np.floor(arr)) # rounds off to higher number 
print(np.ceil(arr)) # rounds off to lower number 
```
